```python
# a) Draw a line in a diagram from position (1, 3) to (2, 10) then to (6, 12) and finally to position (18, 20).
(Mark each point with a beautiful green colour and set line colour to red and line style dotted).
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Define data points
xpoints = np.array([1, 2, 6, 18])
ypoints = np.array([3, 10, 12, 20])
# Create the plot
plt.plot(
    xpoints,
    ypoints,
    marker='o',         # Circle marker
    color='red',        # Line color
    mec='green',        # Marker edge color
    mfc='green',        # Marker face color
    linestyle='dotted'  # Dotted line
)
# Display the plot
plt.show()
```
