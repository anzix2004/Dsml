```python
#### BASIC OPERATIONS
```

```python
#### INDEXING AND SLICING
```

```python
arr=np.array([1,2,3,4,5])
print(arr[0])
#ACCESSING THE FIRST ELEMENT
print(arr[1:3])
#SLICING THE ARRAY
```

```python
#### BASIC MATHEMATICAL OPERATIONS
```

```python
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
#ADDITION
sum_arr=arr1+arr2
print(sum_arr)
 #SUBTRACTION
diff_arr=arr1-arr2
print(diff_arr)
#MULTIPLICATION (ELEMENT-WISE)
prod_arr=arr1*arr2
print(prod_arr)
#DIVISION (ELEMENT-WISE)
div_arr=arr1/arr2
print(div_arr)
```

```python
#### CHECK WHETHER THE ARRAYS ARE EQUAL
```

```python
import numpy as np
arr1=np.array([1,2,3,4])
arr2=np.array([1,2,3,4])
print(np.array_equal(arr1,arr2))
```

```python
####AGGREGATE FUNCTIONS
```

```python
arr=np.array([1,2,3,4,5])
print(np.sum(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
```

```python
#### RESHAPING AND TRANSPOSING
```

```python
arr=np.array([[1,2],[3,4]])
reshaped_arr=arr.reshape(1,4)
print(reshaped_arr)
transposed_arr=np.transpose(arr)
print(transposed_arr)
```
