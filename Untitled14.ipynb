```python
#   Write a Python programming to create a pie chart of the popularity of programming Languages.
#   programming languages : "Java", "Python", "PHP", "JavaScript", "C#", "C++"
#   popularity            : 22.2,     17.6,    8.8,    8,           7.7,   6.7
```

```python
import matplotlib.pyplot as plt
import numpy as np
# Data
y = np.array([22.2, 17.6, 8.8, 8, 7.7, 6.7])
mylabels = ["Java", "Python", "PHP", "JavaScript", "C#", "C++"]
# Create pie chart
plt.pie(y, labels=mylabels)
plt.show()
```

```python
#   Write a Python program to create bar plot scores by group and gender.
#   Use multiple values on the same chart for men and women.
```

```python
import numpy as np
import matplotlib.pyplot as plt
y1 = [22, 30, 35, 35, 26]
y2 = [25, 32, 30, 35, 29]
x_labels = ['G1', 'G2', 'G3', 'G4', 'G5']
x1 = np.arange(len(x_labels))
width = 0.40
# Create grouped bar chart
plt.bar(x1 - 0.2, y1, color='green', width=width, label='Men')
plt.bar(x1 + 0.2, y2, color='red', width=width, label='Women')
# Customize axes and labels
plt.xticks(x1, x_labels)
plt.xlabel("Person")
plt.ylabel("Scores")
plt.title("Scores by Group and Gender")
plt.legend()
plt.show()
```

```python
#  Write a Python program to plot two or more lines on same plot with suitable legends of each line.
```

```python
import matplotlib.pyplot as plt
# Line 1 data
x1 = [10, 20, 30]
y1 = [10, 20, 30]
plt.plot(x1, y1, label="Line 1")
# Line 2 data
x2 = [30, 40, 50]
y2 = [30, 40, 50]
plt.plot(x2, y2, label="Line 2")
# Labels and title
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Two Line Plot')
plt.legend()
plt.show()
```

```python
#  Create a scatter plot using Seaborn featuring functionality to visualize the Iris dataset.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Load dataset
iris = sns.load_dataset("iris")
# Set Seaborn style
sns.set_style("whitegrid")
# Create FacetGrid and plot
sns.FacetGrid(iris, hue="species", height=6) \
    .map(plt.scatter, "sepal_length", "petal_length") \
    .add_legend()
plt.show()
```
