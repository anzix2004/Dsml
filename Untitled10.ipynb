```python
#### CREATING ARRAY USING NUMPY
```

```python
import numpy as np
# create a 2x2 matrix
matrix1 = np.array([[1, 3], [5, 7]])
print("2x2 Matrix:\n", matrix1)
# shape of matrix
print(np.shape(matrix1))
# size of the matrix
print(np.size(matrix1))
```

```python
# create a 3x3 matrix
matrix2 = np.array([[2, 3, 5], [7, 14, 21], [1, 3, 5]])
print("\n3x3 Matrix:\n", matrix2)
# shape of matrix
print(np.shape(matrix2))
# size of the matrix
print(np.size(matrix2))
```

```python
### MATRIX ADDITION AND SUBTRACTION
```

```python
# Matrix addition
matrix1 = np.array([[1, 12], [9, 15]])
matrix2 = np.array([[3, 6], [4, 8]])
result = matrix1 + matrix2
print("Sum =\n", result)
# Matrix subtraction
result = matrix1 - matrix2
print("Difference=\n", result)
```

```python
### PERFORM MATRIX MULTIPLICATION
```

```python
# create two matrices
matrix1 = np.array([[1, 3], [5, 7]])
matrix2 = np.array([[2, 6], [4, 8]])
# calculate the dot product of the two matrices
result = np.dot(matrix1, matrix2)
print("matrix1 x matrix2: \n", result)
```

```python
### TRANSPOSE OF A MATRIX
```

```python
# create a matrix
matrix1 = np.array([[1, 3], [5, 7]])
# get transpose of matrix1
result = np.transpose(matrix1)
print(result)
```

```python
### INVERSE OF A SQUARE MATRIX
```

```python
# create a 3x3 square matrix
matrix1 = np.array([[1, 3, 5], [7, 9, 2], [4, 6, 5]])
# find inverse of matrix1
result = np.linalg.inv(matrix1)
print(result)
```

```python
### DETERMINANT OF A MATRIX
```

```python
# create a matrix
matrix1 = np.array([[1, 2, 3], [4, 5, 1], [2, 3, 4]])
# find determinant of matrix1
result = np.linalg.det(matrix1)
print(result)
```

```python
### SINGULAR VALUE DECOMPOSITION(SVD)
```

```python
matrix = np.array([[1, 2], [3, 4]])
U, s, Vh = np.linalg.svd(matrix)
print(U)
print(s)
print(Vh)
```
