```python
import pandas as pd
from sklearn.datasets import load_iris
# Load the Iris dataset from scikit-learn
iris = load_iris()
# Create the DataFrame
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
# Add the target (species) column
iris_df['species'] = pd.Categorical.from_codes(
    iris.target, iris.target_names
)
# Now you can run your code
print(iris_df.info())
```

```python
#   Check for total number of missing values
```

```python
print(iris_df.isna().sum())
```

```python
### Group by species and calculate mean measurements
```

```python
print(iris_df.groupby('species').mean())
```
